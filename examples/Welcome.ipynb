# Welcome to WellPlay 歡迎使用星泉筆記本

WellPlay runs **AquariusLang** in cells, like JupyterLab. Press `Shift+Enter` to run a cell and move on.

- Variables from cells you ran stay available in later cells.
- `Display` shows rich outputs: Markdown, tables, charts, images and progress.
- A cell's last expression shows as `Out[n]`.

```aquarius
var primes = [];
for (var n = 2; len(primes) < 12; n++) {
    var prime = true;
    for (var d = 2; d * d <= n; d++) { if (n % d == 0) { prime = false; break; } }
    if (prime) { Array.Append(primes, n); }
}
print("The first", len(primes), "primes:");
primes
```

```aquarius
Display.Table(Array.Map(primes, p => ({"prime": p, "square": p * p, "gap": p - 1})))
```

```aquarius
var squares = Array.Map(primes, p => p * p);
Display.Show(Display.LineChart({"primes": primes, "half squares": Array.Map(squares, s => s / 2)}, title: "Primes and their squares", height: 240));
Display.BarChart(primes, title: "Bars", height: 200)
```

## Rich text 豐富文字

> Markdown cells render headings, **bold**, *italic*, `code`, lists and quotes.

1. Double-click a rendered cell to edit it.
2. Press `Esc` for command mode, then `A` / `B` to add cells.

```aquarius
var math = import("Processing");
var points = [];
for (var i = 0; i < 60; i++) { Array.Append(points, [math.cos(i * 0.2d) * i, math.sin(i * 0.2d) * i]); }
Display.ScatterChart([{"name": "spiral", "points": points}], title: "A spiral 螺旋")
```

```aquarius
# An output with an id updates in place while the cell runs. 有 id 的輸出可以在執行中原地更新。
Display.Show(Display.Progress(0, 20, "Working 工作中"), "work");
for (var step = 1; step <= 20; step++) {
    await Promise.Delay(100);
    Display.Update("work", Display.Progress(step, 20, "Working 工作中"));
}
Display.Markdown("**Done!** 完成 ✓")
```
