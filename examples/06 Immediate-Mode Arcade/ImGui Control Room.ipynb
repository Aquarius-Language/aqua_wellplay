# ImGui Control Room ImGui 控制室

Immediate-mode UI rebuilds every frame; `drawUI` paints its draw list.

```aquarius
var ImGui = import("ImGui");
var gui = ImGui.CreateContext();
gui.AddDefaultFont(15);
# Chinese labels need a CJK font from the system. 中文需要系統的中日韓字型。
var cjk = ImGui.SystemFontPath("cjk");
if (cjk != null) { gui.AddFont(cjk, 16, true); }
var speed = 1.5d;
var wave = [];
for (var i = 0; i < 90; i++) { Array.Append(wave, 0.5d + 0.4d * Math.Sin(i * 0.15d)); }
Display.Sketch((p, frame) => {
    gui.NewFrame(p.width, p.height, 1.0d / 60);
    ImGui.SetNextWindowPos(16, 16);
    ImGui.SetNextWindowSize(300, 250);
    if (ImGui.Begin("Reactor 反應爐")) {
        speed = ImGui.SliderFloat("speed", speed, 0, 5);
        ImGui.ProgressBar(0.65d);
        ImGui.PlotLines("##wave", wave, 0, "core temperature", 0, 1, 0, 80);
        ImGui.Text("Status: nominal");
    }
    ImGui.End();
    ImGui.SetNextWindowPos(330, 16);
    ImGui.SetNextWindowSize(220, 180);
    if (ImGui.Begin("Crew")) {
        if (ImGui.BeginTable("crew", 2, ImGui.TableFlags_Borders)) {
            var crew = [["Ada", "pilot"], ["Linus", "engineer"], ["星泉", "navigator"]];
            for (var i = 0; i < len(crew); i++) { ImGui.TableNextRow(); ImGui.TableNextColumn(); ImGui.Text(crew[i][0]); ImGui.TableNextColumn(); ImGui.Text(crew[i][1]); }
            ImGui.EndTable();
        }
    }
    ImGui.End();
    p.background(20, 22, 30);
    p.drawUI(gui.Render());
    return null;
}, 570, 290, frames: 3)
```
