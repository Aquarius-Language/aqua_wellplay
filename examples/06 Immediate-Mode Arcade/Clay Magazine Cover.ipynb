# Clay Magazine Cover Clay 雜誌封面

Clay declares the whole layout every frame; nested `Clay.Element` calls make boxes.

```aquarius
var Clay = import("Clay");
var layout = null;
Display.Sketch((p, frame) => {
    if (layout == null) { layout = Clay.Create(p.width, p.height); }
    layout.SetDimensions(p.width, p.height);
    layout.Begin();
    Clay.Element({"id": "cover", "width": "grow", "height": "grow", "direction": "column", "padding": 28, "gap": 14, "background": "#1d1b26"}, function() {
        Clay.Text("AQUARIUS MONTHLY", {"fontSize": 14, "fontWeight": 800, "color": "#ffd166"});
        Clay.Text("The Well That Runs Code", {"fontSize": 34, "fontWeight": 800, "color": "#f6f4ef"});
        Clay.Element({"width": "grow", "gap": 12}, function() {
            var tags = ["Notebooks", "Saturn", "ImGui", "Clay"];
            for (var i = 0; i < len(tags); i++) {
                Clay.Element({"padding": [6, 12], "radius": 14, "background": "#3ad7ee33"}, function() { Clay.Text(tags[i], {"fontSize": 14, "color": "#3ad7ee"}); });
            }
        });
        Clay.Element({"width": "grow", "height": "grow", "radius": 16, "background": "#a58bff44", "padding": 20, "alignY": "bottom"}, function() {
            Clay.Text("Inside: 22 recipes for joyful scripts 二十二道快樂腳本", {"fontSize": 18, "color": "#f6f4ef"});
        });
    });
    p.background(0);
    p.drawUI(layout.End());
    return null;
}, 520, 340, frames: 2)
```
