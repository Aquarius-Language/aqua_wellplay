# JSON Time Capsule JSON 時光膠囊

`currWorkingDir` is the notebook's own folder, so files land right next to it.

```aquarius
var Files = import("Files");
var Serialization = import("Serialization");
var capsule = {"buried": currentDate(), "message": "Hello, future me 你好，未來的我", "lucky": [7, 21, 42], "mood": "curious"};
var file = Files.Resolve(currWorkingDir + "/time-capsule.json");
Files.SaveText(file, Serialization.Json(capsule), "utf-8");
print("Buried", Files.Stat(file).length, "bytes in time-capsule.json");
```

```aquarius
var found = Serialization.ParseJson(Files.ReadText(file, "utf-8"));
Display.JSON(found)
```
