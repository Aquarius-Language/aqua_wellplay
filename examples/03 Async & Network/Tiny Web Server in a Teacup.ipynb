# Tiny Web Server in a Teacup 茶杯裡的網頁伺服器

The Network library's Go-style server runs right inside the cell.

```aquarius
var Network = import("Network");
var server = Network.Listen("http://127.0.0.1:18777");
server.Handle("GET /fortune/{name}", function(request, writer) {
    var fortunes = ["a bug will fix itself", "your build will be green", "a pull request will be merged"];
    var name = request.Param("name");
    writer.JSON({"name": name, "fortune": fortunes[len(name) % len(fortunes)]});
    return null;
});
server.ServeAsync();
var answers = [];
var people = ["ann", "bo", "星泉"];
for (var i = 0; i < len(people); i++) {
    var response = await Network.GetAsync("http://127.0.0.1:18777/fortune/" + people[i]);
    Array.Append(answers, response.JSON());
}
server.Shutdown();
Display.Table(answers)
```
