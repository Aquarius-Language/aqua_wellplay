# Async Coffee Bar 非同步咖啡吧

Async functions run side by side; `await` waits without blocking the others. Top-level `await` works in cells.

```aquarius
var brew = async function(drink, ms) {
    await Promise.Delay(ms);
    return drink + " ready after " + String.From(ms) + " ms";
};
var started = currentTime();
var orders = await Promise.All([brew("Latte", 400), brew("Matcha", 250), brew("Mocha", 600)]);
print(String.Join(orders, String.FromCode(10)));
print("All served in", Math.Round(currentTime() - started), "ms (not 1250!)")
```

```aquarius
Display.Show(Display.Progress(0, 10, "Pouring 倒咖啡中"), "pour");
for (var i = 1; i <= 10; i++) {
    await Promise.Delay(120);
    Display.Update("pour", Display.Progress(i, 10, if (i == 10) { "Enjoy! 請享用"; } else { "Pouring 倒咖啡中"; }));
}
```
