# Monte Carlo Dartboard 蒙地卡羅飛鏢

Darts that land inside the quarter circle, divided by all darts, times four, approach **π**.

```aquarius
Math.Seed(2026);
var inside = [];
var outside = [];
var estimates = [];
var hits = 0;
for (var i = 1; i <= 1500; i++) {
    var x = Math.Random();
    var y = Math.Random();
    var hit = x * x + y * y <= 1;
    if (hit) { hits++; }
    if (i <= 600) { Array.Append(if (hit) { inside; } else { outside; }, [x, y]); }
    if (i % 25 == 0) { Array.Append(estimates, 4.0d * hits / i); }
}
Display.ScatterChart([{"name": "inside", "points": inside}, {"name": "outside", "points": outside}], title: "600 darts", height: 320)
```

```aquarius
Display.Show(Display.LineChart({"estimate": estimates, "π": Array.Map(estimates, e => Math.PI)}, title: "Closing in on π 逼近圓周率", height: 240));
print("Final estimate:", estimates[len(estimates) - 1], "error:", Math.Abs(estimates[len(estimates) - 1] - Math.PI))
```
