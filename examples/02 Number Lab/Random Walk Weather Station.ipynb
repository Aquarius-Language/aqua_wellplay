# Random Walk Weather Station 隨機漫步氣象站

Each day the temperature drifts a little. A seed makes the weather repeatable.

```aquarius
Math.Seed(7);
var cities = {"Taipei": 24.0d, "Reykjavik": 4.0d, "Cairo": 30.0d, "Lima": 18.0d};
var season = {};
var names = Hash.Keys(cities);
for (var c = 0; c < len(names); c++) {
    var t = cities[names[c]];
    var days = [];
    for (var d = 0; d < 30; d++) { t += (Math.Random() - 0.5d) * 3; Array.Append(days, Math.Round(t * 10) / 10); }
    season[names[c]] = days;
}
Display.AreaChart(season, title: "Thirty days of weather 三十天天氣", height: 300)
```

```aquarius
Display.Table(Array.Map(names, n => ({"city": n, "coldest": Array.Min(season[n]), "warmest": Array.Max(season[n]), "average": Math.Round(Array.Sum(season[n]) / 30 * 10) / 10})))
```
