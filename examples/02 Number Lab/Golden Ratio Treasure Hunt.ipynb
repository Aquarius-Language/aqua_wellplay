# Golden Ratio Treasure Hunt 黃金比例尋寶

The ratio of neighboring Fibonacci numbers approaches **φ = 1.6180339…**

```aquarius
var fib = [1L, 1L];
for (var i = 2; i < 40; i++) { Array.Append(fib, fib[i - 1] + fib[i - 2]); }
var phi = (1 + Math.Sqrt(5)) / 2;
var ratios = [];
for (var i = 1; i < 20; i++) { Array.Append(ratios, double(fib[i]) / double(fib[i - 1])); }
Display.LineChart({"F(n)/F(n-1)": ratios, "φ": Array.Map(ratios, r => phi)}, title: "Racing to φ", height: 240)
```
