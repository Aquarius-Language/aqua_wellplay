# Sorting Hat Race 排序大賽

Three sorts, one shuffled array, a stopwatch.

```aquarius
Math.Seed(42);
var deck = [];
for (var i = 0; i < 1500; i++) { Array.Append(deck, Math.RandomInt(0, 100000)); }
var bubble = function(a) {
    var n = len(a);
    for (var i = 0; i < n; i++) { for (var j = 0; j + 1 < n - i; j++) { if (a[j] > a[j + 1]) { var t = a[j]; a[j] = a[j + 1]; a[j + 1] = t; } } }
    return a;
};
var insertion = function(a) {
    for (var i = 1; i < len(a); i++) {
        var v = a[i];
        var j = i - 1;
        for (; j >= 0 && a[j] > v; j -= 1) { a[j + 1] = a[j]; }
        a[j + 1] = v;
    }
    return a;
};
var race = {};
var contestants = {"bubble": bubble, "insertion": insertion, "Array.Sort": a => Array.Sort(a)};
var names = Hash.Keys(contestants);
for (var k = 0; k < len(names); k++) {
    var copy = Array.Clone(deck);
    var started = currentTime();
    contestants[names[k]](copy);
    race[names[k]] = Math.Round(currentTime() - started);
}
Display.BarChart(Array.Map(names, n => race[n]), title: "Milliseconds to sort 1,500 cards", labels: names, height: 220)
```
