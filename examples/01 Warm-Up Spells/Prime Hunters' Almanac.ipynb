# Prime Hunters' Almanac 質數獵人年鑑

The sieve of Eratosthenes, then a look at the **gaps** between primes. Run the cells in order with `Shift+Enter`.

```aquarius
var limit = 2000;
var composite = Array.Make(limit + 1, false);
var primes = [];
for (var n = 2; n <= limit; n++) {
    if (composite[n]) { continue; }
    Array.Append(primes, n);
    for (var m = n * n; m <= limit; m += n) { composite[m] = true; }
}
print("Found", len(primes), "primes below", limit);
Array.Slice(primes, 0, 20)
```

```aquarius
var gaps = [];
for (var i = 1; i < len(primes); i++) { Array.Append(gaps, primes[i] - primes[i - 1]); }
Display.BarChart(Array.Slice(gaps, 0, 60), title: "The first 60 prime gaps 質數間距", height: 220)
```

```aquarius
# Twin primes: pairs that differ by 2. 孿生質數
var twins = [];
for (var i = 1; i < len(primes); i++) { if (primes[i] - primes[i - 1] == 2) { Array.Append(twins, {"smaller": primes[i - 1], "larger": primes[i]}); } }
Display.Show(Display.Markdown("There are **" + String.From(len(twins)) + "** twin prime pairs below " + String.From(limit) + "."));
Display.Table(Array.Slice(twins, len(twins) - 8, len(twins)))
```
