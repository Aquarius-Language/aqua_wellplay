# Bitwise Bakery 位元烘焙坊

Integers are 32-bit; `&`, `|`, `^`, `~`, `<<`, `>>` and `>>>` work on their bits.

```aquarius
var binary = function(n, width = 8) {
    var text = "";
    for (var bit = width - 1; bit >= 0; bit -= 1) { text += if (((n >> bit) & 1) == 1) { "1"; } else { "0"; }; }
    return text;
};
var dough = 0b10110010;
Display.Table([
    {"step": "dough", "bits": binary(dough), "value": dough},
    {"step": "fold left (<< 1)", "bits": binary((dough << 1) & 255), "value": (dough << 1) & 255},
    {"step": "rest (>> 2)", "bits": binary(dough >> 2), "value": dough >> 2},
    {"step": "sprinkle (| 0x0F)", "bits": binary(dough | 0x0F), "value": dough | 0x0F},
    {"step": "glaze (^ 0xFF)", "bits": binary(dough ^ 0xFF), "value": dough ^ 0xFF},
])
```

```aquarius
# Count the set bits of every byte. 每個位元組有幾個 1？
var ones = n => { var c = 0; for (var v = n; v != 0; v = v & (v - 1)) { c++; } return c; };
Display.BarChart(Array.Map(Array.Make(64, 0), (x) => 0), height: 10);
var counts = [];
for (var b = 0; b < 64; b++) { Array.Append(counts, ones(b)); }
Display.BarChart(counts, title: "Set bits in 0..63 位元計數", height: 220)
```
