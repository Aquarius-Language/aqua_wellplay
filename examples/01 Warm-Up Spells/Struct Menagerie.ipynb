# Struct Menagerie 結構動物園

Structs hold fields (with defaults), functions take named arguments, and `x => ...` writes short functions.

```aquarius
struct Creature { name; habitat = "forest"; legs = 4; snack = "berries"; weight = 1.0d }
var zoo = [
    Creature("Fennec fox", habitat: "desert", weight: 1.3d),
    Creature("Octopus", habitat: "reef", legs: 8, snack: "crabs", weight: 15.0d),
    Creature("Kiwi", legs: 2, snack: "worms", weight: 2.4d),
    Creature("Pangolin", snack: "ants", weight: 12.0d),
    Creature("Axolotl", habitat: "lake", snack: "shrimp", weight: 0.2d),
];
Display.Table(Array.Map(zoo, c => ({"name": c.name, "habitat": c.habitat, "legs": c.legs, "snack": c.snack, "kg": c.weight})))
```

```aquarius
var describe = (c, loud = false) => {
    var line = c.name + " lives in the " + c.habitat + " and loves " + c.snack;
    return if (loud) { String.Upper(line) + "!"; } else { line + "."; };
};
for (var i = 0; i < len(zoo); i++) { print(describe(zoo[i], loud: zoo[i].legs > 4)); }
```

```aquarius
Display.BarChart({"weight (kg)": Array.Map(zoo, c => c.weight)}, title: "Who is heaviest? 誰最重？", labels: Array.Map(zoo, c => c.name), height: 240)
```
