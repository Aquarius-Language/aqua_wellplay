# Word Alchemist 文字煉金師

Strings, hashes and sorting turn a poem into numbers.

```aquarius
var NL = String.FromCode(10);
var poem = String.Join([
    "the well is deep and the water is bright",
    "drop a stone and wait for the light",
    "星泉深深 水光閃閃",
    "the stars drink deep from the well tonight",
], NL);
var counts = {};
var words = String.Split(String.Replace(poem, NL, " "), " ");
for (var i = 0; i < len(words); i++) {
    var w = String.Lower(words[i]);
    if (w == "") { continue; }
    counts[w] = Hash.Get(counts, w, 0) + 1;
}
var ranked = Hash.Keys(counts);
Array.Sort(ranked, (a, b) => counts[b] - counts[a]);
Display.Table(Array.Map(Array.Slice(ranked, 0, 8), w => ({"word": w, "count": counts[w]})))
```

```aquarius
var letters = {};
for (var i = 0; i < len(poem); i++) {
    var c = poem[i];
    var code = String.Code(c, 0);
    if (code >= 97 && code <= 122) { letters[c] = Hash.Get(letters, c, 0) + 1; }
}
var keys = Array.Sort(Hash.Keys(letters));
Display.BarChart(Array.Map(keys, k => letters[k]), title: "Letter frequency 字母頻率", labels: keys, height: 220)
```
