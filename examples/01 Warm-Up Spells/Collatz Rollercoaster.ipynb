# Collatz Rollercoaster 考拉茲雲霄飛車

Halve even numbers, triple-and-add-one odd ones, and every start seems to reach **1**. Nobody has proven it.

```aquarius
var hailstone = function(start) {
    var path = [start];
    for (var n = start; n != 1;) {
        n = if (n % 2 == 0) { n / 2; } else { 3 * n + 1; };
        Array.Append(path, n);
    }
    return path;
};
Array.Slice(hailstone(27), 0, 12)
```

```aquarius
Display.LineChart({"27": hailstone(27), "97": hailstone(97), "871": hailstone(871)}, title: "Three rides 三趟旅程", height: 300)
```

```aquarius
# Who takes the longest ride below 3000? 誰坐最久？
var best = 1;
var longest = 1;
for (var s = 1; s < 3000; s++) {
    var steps = len(hailstone(s));
    if (steps > longest) { longest = steps; best = s; }
}
Display.Markdown("**" + String.From(best) + "** rides for **" + String.From(longest) + "** stops.")
```
