# Neon Orbit Postcard 霓虹軌道明信片

`Display.Sketch(draw, width, height)` runs a Processing sketch on this computer and shows its last frame in the cell.

```aquarius
Display.Sketch((p, frame) => {
    p.background(8, 10, 28);
    p.noFill();
    p.translate(p.width / 2, p.height / 2);
    for (var ring = 1; ring <= 5; ring++) {
        p.stroke(60 + ring * 30, 120, 255, 90);
        p.ellipse(0, 0, ring * 90, ring * 52);
        var a = ring * 1.3d;
        p.noStroke();
        for (var glow = 5; glow > 0; glow -= 1) {
            p.fill(255 - ring * 30, 80 + ring * 30, 220, 40);
            p.ellipse(p.cos(a) * ring * 45, p.sin(a) * ring * 26, glow * 7, glow * 7);
        }
        p.fill(255);
        p.ellipse(p.cos(a) * ring * 45, p.sin(a) * ring * 26, 8, 8);
        p.noFill();
    }
    p.fill(255, 210, 90);
    p.ellipse(0, 0, 34, 34);
    return null;
}, 560, 320)
```
