# Bouncing Comet Window 彈跳彗星視窗

This cell opens a **real window on the computer running WellPlay**. Move the mouse to pull the comet; close the window (or press Stop) to end the cell.

```aquarius
var p = import("Processing");
var x = 200.0d;
var y = 150.0d;
var vx = 3.0d;
var vy = 2.0d;
p.run(function() { p.size(480, 320, p.P2D, "Bouncing Comet 彈跳彗星"); return null; }, function() {
    p.fill(6, 8, 24, 40);
    p.rect(0, 0, p.width, p.height);
    vx += (p.mouseX - x) * 0.0008d;
    vy += (p.mouseY - y) * 0.0008d;
    x += vx;
    y += vy;
    if (x < 10 || x > p.width - 10) { vx = -vx; }
    if (y < 10 || y > p.height - 10) { vy = -vy; }
    p.noStroke();
    p.fill(255, 200, 90);
    p.ellipse(x, y, 18, 18);
    return null;
});
print("The window closed. 視窗已關閉。");
```
