# Mandelbrot Snow Globe 曼德博雪花球

No canvas needed: the Images module turns raw RGBA bytes into a picture.

```aquarius
var Images = import("Images");
var w = 360;
var h = 240;
var pixels = [];
for (var y = 0; y < h; y++) {
    for (var x = 0; x < w; x++) {
        var cr = -2.2d + 3.0d * x / w;
        var ci = -1.0d + 2.0d * y / h;
        var zr = 0.0d;
        var zi = 0.0d;
        var n = 0;
        for (; n < 60 && zr * zr + zi * zi < 4; n++) {
            var t = zr * zr - zi * zi + cr;
            zi = 2 * zr * zi + ci;
            zr = t;
        }
        if (n == 60) { Array.AppendArray(pixels, [10, 12, 30, 255]); }
        else { Array.AppendArray(pixels, [Math.Min(255, n * 9), Math.Min(255, 40 + n * 5), 255 - n * 3, 255]); }
    }
}
Display.Image(Images.Create(w, h, pixels))
```
