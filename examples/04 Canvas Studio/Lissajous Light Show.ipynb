# Lissajous Light Show 利薩茹光秀

With `live`, the sketch updates the cell every few frames: an inline animation.

```aquarius
var a = 3;
var b = 4;
Display.Sketch((p, frame) => {
    p.background(5, 8, 20);
    p.translate(p.width / 2, p.height / 2);
    p.noFill();
    p.strokeWeight(2);
    var t = frame * 0.04d;
    for (var i = 0; i < 200; i++) {
        var s = t - i * 0.015d;
        p.stroke(80 + i, 200 - i / 2, 255, 255 - i);
        p.point(p.sin(a * s + 1.57d) * 200, p.sin(b * s) * 120);
    }
    return null;
}, 520, 300, frames: 120, live: 6)
```
