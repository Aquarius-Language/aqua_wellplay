# Particle Fountain 粒子噴泉

State lives in arrays outside the sketch; every frame moves it a little.

```aquarius
struct Particle { x; y; vx; vy; life }
Math.Seed(3);
var particles = [];
Display.Sketch((p, frame) => {
    for (var k = 0; k < 8; k++) {
        Array.Append(particles, Particle(260.0d, 300.0d, (Math.Random() - 0.5d) * 4, -6 - Math.Random() * 4, 1.0d));
    }
    p.background(12, 14, 30);
    p.noStroke();
    var alive = [];
    for (var i = 0; i < len(particles); i++) {
        var q = particles[i];
        q.vy += 0.18d;
        q.x += q.vx;
        q.y += q.vy;
        q.life -= 0.012d;
        if (q.life > 0 && q.y < 320) {
            p.fill(80 + q.life * 170, 160, 255, q.life * 255);
            p.ellipse(q.x, q.y, 6, 6);
            Array.Append(alive, q);
        }
    }
    particles = alive;
    return null;
}, 520, 320, frames: 150, live: 5)
```
