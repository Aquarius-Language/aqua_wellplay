# Pocket Counter App 口袋計數器

`UI.RunApp` opens an interactive window on this computer. Close it to finish the cell.

```aquarius
var UI = import("SaturnUI");
struct CounterState { count = 0; history = [] }
var app = UI.Stateful(
    init: () => CounterState(),
    build: (state, context) => UI.Column(gap: 12, crossAxis: "center", style: {"padding": 24}, children: [
        UI.Text(String.From(state.count), style: {"fontSize": 48, "fontWeight": 800}),
        UI.Row(gap: 8, children: [
            UI.Button("+1", onPress: () => { state.count += 1; Array.Append(state.history, "+1 → " + String.From(state.count)); }),
            UI.Button("Reset", variant: "outline", onPress: () => { state.count = 0; state.history = []; }),
        ]),
        for (var i = 0; i < len(state.history) && i < 5; i++) UI.Text(state.history[len(state.history) - 1 - i], style: {"color": "#8b9ab3"}),
    ]),
);
UI.RunApp(app, {"title": "Pocket Counter", "width": 360, "height": 420, "theme": UI.Theme(mode: "dark")});
```
