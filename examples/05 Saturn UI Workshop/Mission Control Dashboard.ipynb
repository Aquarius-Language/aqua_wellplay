# Mission Control Dashboard 任務控制儀表板

`Display.UI(widget, width, height, dark)` lays out a SaturnUI tree with Yoga and shows it inline.

```aquarius
var UI = import("SaturnUI");
var stat = (label, value, tone) => UI.Card(children: [
    UI.Text(label, style: {"fontSize": 12, "color": "#8b9ab3"}),
    UI.Text(value, style: {"fontSize": 26, "fontWeight": 800, "color": tone}),
]);
Display.UI(UI.Column(gap: 14, crossAxis: "stretch", style: {"padding": 20}, children: [
    UI.Text("Mission Control 任務控制", style: {"fontSize": 22, "fontWeight": 800}),
    UI.Row(gap: 12, children: [
        UI.Expanded(child: stat("Altitude", "412 km", "#3ad7ee")),
        UI.Expanded(child: stat("Speed", "7.66 km/s", "#a58bff")),
        UI.Expanded(child: stat("Crew", "7", "#5ee6a0")),
    ]),
    UI.ProgressBar(value: 0.72d, width: 520),
    UI.Row(gap: 8, children: [UI.Button("Launch", icon: "play"), UI.Button("Abort", variant: "danger"), UI.Button("Logs", variant: "outline")]),
]), 560, 260, dark: true)
```
