# Theme Swatch Gallery 主題色票館

Themes recolor every widget; a custom `primary` is one argument away.

```aquarius
var UI = import("SaturnUI");
var sample = (name, theme) => UI.Themed(theme: theme, child: UI.Column(gap: 10, style: {"padding": 16, "borderRadius": 12}, children: [
    UI.Text(name, style: {"fontWeight": 800, "fontSize": 16}),
    UI.Button("Primary"),
    UI.Button("Outline", variant: "outline"),
    UI.Switch(value: true, label: "Switch"),
    UI.Slider(value: 0.6d, width: 140),
]));
Display.UI(UI.Row(gap: 0, crossAxis: "stretch", style: {"flexGrow": 1}, children: [
    UI.Expanded(child: sample("Light", UI.Theme(mode: "light"))),
    UI.Expanded(child: sample("Dark", UI.Theme(mode: "dark"))),
    UI.Expanded(child: sample("Sunset", UI.Theme(mode: "dark", primary: "#ff7a59", background: "#2a1633"))),
]), 600, 250)
```
