# Game of Life Lava Lamp 生命遊戲熔岩燈

Each generation becomes an image; `Display.Update` swaps it in place.

```aquarius
var Images = import("Images");
var w = 64;
var h = 40;
var scale = 6;
Math.Seed(11);
var grid = [];
for (var i = 0; i < w * h; i++) { Array.Append(grid, Math.Random() < 0.3d); }
var picture = function(cells, generation) {
    var pixels = [];
    for (var y = 0; y < h * scale; y++) {
        for (var x = 0; x < w * scale; x++) {
            var on = cells[int(y / scale) * w + int(x / scale)];
            if (on) { Array.AppendArray(pixels, [255, 120 + (generation * 3) % 120, 60, 255]); } else { Array.AppendArray(pixels, [20, 10, 40, 255]); }
        }
    }
    return Display.Image(Images.Create(w * scale, h * scale, pixels));
};
var step = function(cells) {
    var next = [];
    for (var y = 0; y < h; y++) {
        for (var x = 0; x < w; x++) {
            var n = 0;
            for (var dy = -1; dy <= 1; dy++) { for (var dx = -1; dx <= 1; dx++) {
                if (dx == 0 && dy == 0) { continue; }
                if (cells[((y + dy + h) % h) * w + (x + dx + w) % w]) { n++; }
            } }
            var alive = cells[y * w + x];
            Array.Append(next, (alive && (n == 2 || n == 3)) || (!alive && n == 3));
        }
    }
    return next;
};
Display.Show(picture(grid, 0), "life");
for (var g = 1; g <= 25; g++) { grid = step(grid); Display.Update("life", picture(grid, g)); }
```
