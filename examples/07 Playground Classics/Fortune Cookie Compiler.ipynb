# Fortune Cookie Compiler 幸運餅乾編譯器

Run the cell again for a new fortune (change the seed, or remove `Math.Seed`).

```aquarius
Math.Seed(int(currentTime()) % 1000);
var luck = ["great", "good", "curious", "surprising", "legendary"];
var omens = ["a forgotten semicolon", "a green build", "a rubber duck", "a merge without conflicts", "a mysterious off-by-one"];
var advice = ["write the test first", "take a walk", "read the error message twice", "refactor gently", "ship it"];
var pick = list => list[Math.RandomInt(0, len(list))];
var NL = String.FromCode(10);
Display.Markdown("## Today's fortune 今日運勢" + NL + NL + "Your day will be **" + pick(luck) + "**. Watch for *" + pick(omens) + "*." + NL + NL + "> Advice: " + pick(advice) + ".")
```
