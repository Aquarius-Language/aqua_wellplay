# Maze Weaver 迷宮織者

A randomized depth-first search visits every room once: one path between any two rooms.

```aquarius
var Images = import("Images");
var cols = 24;
var rows = 14;
var cell = 14;
Math.Seed(99);
var width = cols * 2 + 1;
var height = rows * 2 + 1;
var open = Array.Make(width * height, false);
var visited = Array.Make(cols * rows, false);
var stack = [[0, 0]];
visited[0] = true;
open[width + 1] = true;
for (; len(stack) > 0;) {
    var top = stack[len(stack) - 1];
    var options = [];
    var moves = [[1, 0], [-1, 0], [0, 1], [0, -1]];
    for (var m = 0; m < 4; m++) {
        var nx = top[0] + moves[m][0];
        var ny = top[1] + moves[m][1];
        if (nx >= 0 && ny >= 0 && nx < cols && ny < rows && !visited[ny * cols + nx]) { Array.Append(options, [nx, ny]); }
    }
    if (len(options) == 0) { Array.Delete(stack, len(stack) - 1, len(stack)); continue; }
    var next = options[Math.RandomInt(0, len(options))];
    visited[next[1] * cols + next[0]] = true;
    open[(next[1] * 2 + 1) * width + next[0] * 2 + 1] = true;
    open[(top[1] + next[1] + 1) * width + top[0] + next[0] + 1] = true;
    Array.Append(stack, next);
}
var pixels = [];
for (var y = 0; y < height * cell / 2; y++) {
    for (var x = 0; x < width * cell / 2; x++) {
        var wall = !open[int(y * 2 / cell) * width + int(x * 2 / cell)];
        Array.AppendArray(pixels, if (wall) { [30, 34, 60, 255]; } else { [240, 230, 200, 255]; });
    }
}
Display.Image(Images.Create(width * cell / 2, height * cell / 2, pixels))
```
